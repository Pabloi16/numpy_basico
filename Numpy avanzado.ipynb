```python
import numpy as np
import ssl

# 1. Saltamos la restricción de seguridad SSL que te dio problemas antes
ssl._create_default_https_context = ssl._create_unverified_context

# 2. Definimos la URL correcta (asegúrate de que sea esta)
ruta_url = 'https://raw.githubusercontent.com/selva86/datasets/master/Auto.csv'

# 3. Importamos los datos directamente de la web
# Usamos dtype='float' para manejar los decimales (float64 por defecto)
data3 = np.genfromtxt(ruta_url, delimiter=',', skip_header=1, filling_values=-999, dtype='float')

# 4. Verificamos que se cargó viendo las primeras 3 filas
print("Datos cargados correctamente:")
print(data3[:3])
```

```python
# Guardamos el arreglo que está en memoria a un archivo físico en tu computadora
np.savetxt("out.csv", data3, delimiter=",")

print("¡Archivo 'out.csv' creado exitosamente!")
```
